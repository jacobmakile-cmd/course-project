## Description

This notebook contains a Python program that:
1. Converts temperature from Celsius to Fahrenheit
2. Reports whether the temperature is above or below a threshold
3. Uses a loop to convert several temperatures
4. Uses a function for the conversion

```python
# ============================================
# ASSIGNMENT 2: TEMPERATURE CONVERTER
# ============================================

# --- Function Definition ---
def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit using F = (C × 9/5) + 32"""
    return (celsius * 9/5) + 32

# --- Threshold ---
THRESHOLD = 25  # °C

# --- Main Program ---
print("=" * 50)
print("TEMPERATURE CONVERTER")
print("=" * 50)
print(f"Threshold: {THRESHOLD}°C\n")

# Ask how many temperatures
n = int(input("How many temperatures? "))

# Loop through temperatures
for i in range(n):
    print(f"\n--- Temperature {i+1} ---")
    c = float(input("Enter temperature in Celsius: "))
    f = celsius_to_fahrenheit(c)
    print(f"{c}°C = {f:.2f}°F")

    if c > THRESHOLD:
        print(f"ABOVE threshold ({THRESHOLD}°C)")
    elif c < THRESHOLD:
        print(f"BELOW threshold ({THRESHOLD}°C)")
    else:
        print(f"EQUAL to threshold ({THRESHOLD}°C)")

print("\n" + "=" * 50)
print("Done!")
```
